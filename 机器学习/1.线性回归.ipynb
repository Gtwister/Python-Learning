## 线性回归

$$
\large y =\theta_0 + \theta_1x_1 +  \theta_2x_2 \dots + \theta_kx_k \quad (k=1,2,3,\dots)
$$


* 假设$\theta_1$是年龄的参数,$\theta_2$是工资的参数
* 拟合的平面:$h_{\theta}(x)=\theta_0 + \theta_1x_1 +  \theta_2x_2 \quad$($\theta_0$是偏置项)
* 整合:$h_{\theta}(x)=\sum_{i=0}^{n}\theta_ix_i$=$\theta^{T}x$

**误差**
* 真实值和预测值之间肯定是要存在差异的(用$\varepsilon$来表示误差)
* 对于每个样本: $y^{(i)}=\theta^{T}x^{(i)}+\varepsilon^{(i)}\quad(i代表不同的样本)$ （1）
* 误差$\varepsilon^{(i)}$是独立并且具有相同的分布,并且服从均值为0方差为$\theta^2$的高斯分布

**高斯分布(正态分布)**
* $\large p(\epsilon^{(i)})=\frac{1}{\sqrt{2\pi}\sigma}\exp^{\large-\frac{(\epsilon^{(i)})^2}{2\sigma^{2}}} \quad$ (2)

**(1)式代入(2)式**
* $\large p(y^{(i)}|x^{(i)},\theta)=\frac{1}{\sqrt{2\pi}\sigma}\exp^{-\frac{(y^{(i)-\theta^{T}x^{(i)}})^2}{2\sigma^{2}}}$

**似然函数**
* $\large L(\theta) = \prod_{i=1}^{m}p(y^{(i)}|x^{(i)},\theta) =\prod_{i=1}^{m}\frac{1}{\sqrt{2\pi}\sigma}\exp^{-\frac{(y^{(i)-\theta^{T}x^{(i)}})^2}{2\sigma^{2}}} $

**对数似然**
* $\large \ln L(\theta) = \ln\prod_{i=1}^{m}p(y^{(i)}|x^{(i)},\theta) =\sum_{i=1}^{m}\ln\frac{1}{\sqrt{2\pi}\sigma}\exp^{-\frac{(y^{(i)-\theta^{T}x^{(i)}})^2}{2\sigma^{2}}} $

**展开化简**
* $\large \sum_{i=1}^{m}\ln\frac{1}{\sqrt{2\pi}\sigma}\exp^{-\frac{(y^{(i)-\theta^{T}x^{(i)}})^2}{2\sigma^{2}}} = m\ln\frac{1}{\sqrt{2\pi}\theta} - \frac{1}{\theta^2}\frac{1}{2}\sum_{i=1}^m(y^{(i)}-\theta^{T}x^{(i)}) $
  

    


    
* 要使函数最大,则等式参数部分则要最小
* 即 $\large J(\theta) = \frac{1}{2}\sum_{i=1}^m(y^{(i)}-\theta^{T}x^{(i)})^2\quad(最小二乘)$ 最小

**求解目标函数**
* $ J(\theta) = \frac{1}{2}\sum_{i=1}^m(y^{(i)}-\theta^{T}x^{(i)}) = \frac{1}{2}(X\theta-y)^{T}(X\theta-y)$ $\quad$ $(ab)^{T}=b^Ta^T$

* 求偏导:$\quad$ $\nabla_\theta J(\theta)=\nabla_\theta\frac{1}{2}(X\theta-y)^{T}(X\theta-y)=X^TX\theta-X^Ty$（对参数求偏导等于对参数的转置求导）
* 偏导等于0:$\quad$ $\theta=(X^TX)^{-1}X^Ty$

* 一些细节: $\nabla\theta^{T} A \theta=2A\theta \quad$ (其中$A$是对称阵,$X^TX$也是对称阵)

**梯度下降通俗解释**
* 在loss函数递减的过程中,每次递减我都要重新选择下降速率最大的方向进行减少,这样才能最快达到我想要的最低loss值

**批量梯度下降**
* 目标函数: $\quad J(\theta) = \frac{1}{2}\sum_{i=1}^m(y^{(i)}-\theta^{T}x^{(i)})^2$
* 对目标函数一个样本方向求偏导: $\large\frac{\large\delta J(\theta)}{\large\delta \theta_j}  = -\frac{1}{m}\sum_{i=1}^{m}(y^{i}-h_{\theta}(x^i))x_j^i$
* 因为是梯度下降，所以算出的梯度仍然要反向，所以偏导结果加个负号
* 即可得: $\quad$ $ \theta_j^{new} = \theta_j + \frac{1}{m}\sum_{i=1}^{m}(y^{i}-h_{\theta}(x^i))x_j^i$
* 批量梯度容易得到最优解,但是由于每次考虑所有样本,速度很慢

**梯度下降算法**
$$
\large \theta = \theta_0 - \eta · \nabla f (\theta_0)
$$

**随机梯度下降**
* $$\theta_j^{new} = \theta_j+ (y^i-h_{\theta}(x^i))x_j^i$$
* 每次找一个样本,迭代速度块,但不一定每次都朝着收敛方向

**小批量梯度下降法**
* $$\theta_j^{new} = \theta_j - \alpha\frac{1}{10}\sum_{k=i}^{i+9}(h_{\theta}(x^{(k)})-y^{(k)})x_j^{k} $$
* (每次更新选择一小部分数据来算,实用!)
* 每次迭代的样本量叫做batch
* batch大速度就越慢但结果越精确,batch小速度快但是结果可能比较差

**学习率(learning rate)**
* 对结果会产生巨大影响,一般小一些。
