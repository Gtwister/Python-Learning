## 整体布局风格设置

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sinplot(flip=1):
    x = np.linspace(0,14,100)
    for i in range(1,7):
        plt.plot(x,np.sin(x+i*0.5)*(7-i)*flip)
```

```python
sinplot()
```

```python
sns.set()# 使用seaborn库
sinplot()
```

**五种主题风格**
* darkgrid
* whitegrid
* dark
* white
* ticks

```python
sns.set_style('whitegrid')
data = np.random.normal(size=(20,6))+np.arange(6)/2
sns.boxplot(data=data)
```

```python
sns.set_style('dark')
sinplot()
```

```python
sns.set_style('ticks')
sinplot()
```

```python
sinplot()
sns.despine()# 去掉上方和右方的轴
```

```python
#f,ax = plt.subplots()
sns.violinplot(data)
sns.despine(offset=10)#画的轴离图的距离
```

```python
sns.set_style('whitegrid')
sns.boxplot(data = data,palette='deep')
sns.despine(left = True)#隐藏左边的轴
```

```python
with sns.axes_style('whitegrid'):
    plt.subplot(211)
    sinplot()
plt.subplot(212)
sinplot(-1)
```
