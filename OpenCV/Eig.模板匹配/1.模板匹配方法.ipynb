### 模板匹配方法

matchTemplate( )方法 模板匹配方法

`result = cv2.matchTemplate(image, templ, method, mask)`\
参数说明:\
templ:模板图像,尺寸必须小于或者等于原始图像。\
method:可匹配的方法,见下表


参数值|值|含义
--|--|--
cv2.TM_SQDIFF|0|差值平方和匹配，也叫平方差匹配。可以理解为差异程度。匹配程度越高，计算结果也就越小。完全匹配的结果为0
cv2.TM_SQDIFF_NORMED|1|标准差值平方和匹配,也叫标准平方差匹配,规则同上
cv2.TM_CCORR|2|相关匹配。可以理解为相似程度，匹配程度越高，计算结果越大
cv2.TM_CCORR_NORMED|3|标准相关匹配，规则同上。
cv2.TM_CCOEFF|4|相关系数匹配，也属于相似程度，计算结果为-1~1之间的浮点数，1表示完全匹配，0表示毫无关系,-1表示两张图片亮度刚好相反。
cv2.TM_CCOEFF_NORMED|5|标准相关系数匹配，规则同上
