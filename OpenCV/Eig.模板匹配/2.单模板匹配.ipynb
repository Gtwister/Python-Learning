### 单目标匹配

matchTemplate( )方法计算的是一个二维数组,opencv提供了一个minMaxLoc()的方法专门来解析这个二维数组中的最大值和最小值以及这两个值对应的坐标

`minValue, maxValue, minLoc, maxLoc = cv2.minMaxLoc(src, mask)`

平方差匹配结果越小，匹配程度越高。minMaxLoc()方法返回的minValue值就是模板匹配的最优结果,minLoc就是最优的左上角的点坐标。

```python
import  cv2

img = cv2.imread('../multisubjects.png')
templ = cv2.imread('../heart.png')
height, width, c = templ.shape
results = cv2.matchTemplate(img, templ, cv2.TM_SQDIFF_NORMED) #按照平方差标准方式匹配
# 获取匹配结果中的最大值，最小值，最小值坐标和最大值的坐标
minValue, maxValue, minLoc, maxLoc = cv2.minMaxLoc(results)
resultPoint= (minLoc[0] + width, minLoc[1]+height)
cv2.rectangle(img,minLoc,resultPoint,(0,0,255),2)
cv2.imshow('img',img)
cv2.waitKey()
cv2.destroyAllWindows()
```

### 多目标匹配

```python
import  cv2

img = cv2.imread('../multisubjects.png')
templ = cv2.imread('../pig.jpg')
height, width, c = templ.shape
results = cv2.matchTemplate(img, templ, cv2.TM_SQDIFF_NORMED) #按照平方差标准方式匹配
# 获取匹配结果中的最大值，最小值，最小值坐标和最大值的坐标
for y in range(len(results)): # 遍历结果的列
    for x in range(len(results[y])): # 遍历结果的行
        if results[y][x]>0.3:  
            cv2.rectangle(img,(x,y),(x+width,y+height),(0,0,255),2)
cv2.imshow('img',img)
cv2.waitKey()
cv2.destroyAllWindows()
```
