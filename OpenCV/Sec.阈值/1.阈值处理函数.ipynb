## 阈值

threshold()方法\
`retval, dst = cv2.threshold(src, thresh, maxval, type)`\
参数说明:\
src: 被处理函数，可以是多通道图像\
thresh: 阈值，阈值在125~150范围的取值的效果最好\
maxval: 阈值处理采用的最大值\
type: 阈值处理类型\
返回值说明：\
retval: 处理时采用的阈值\
dst: 经过阈值处理后的图像


**阈值处理类型及其含义**
类型|含义
--|--
`cv2.THRESH_BINARY`|二值化阈值处理
`cv2.THRESH_BINARY_INV`|反二值化阈值处理
`cv2.THRESH_TOZERO`|低于阈值零处理
`cv2.THRESH_TOZERO_INV`|超出阈值零处理
`cv2.THRESH_TRUNC`|截断阈值处理

```python
import cv2
image = cv2.imread("../1.2.jpg")
image_Gray = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
_, dst1= cv2.threshold(image_Gray,127,255,cv2.THRESH_BINARY)
_, dst2= cv2.threshold(image_Gray,127,255,cv2.THRESH_BINARY_INV)
_, dst3= cv2.threshold(image_Gray,127,255,cv2.THRESH_TOZERO)
_, dst4= cv2.threshold(image_Gray,127,255,cv2.THRESH_TOZERO_INV)
_, dst5= cv2.threshold(image_Gray,127,255,cv2.THRESH_TRUNC)
cv2.imshow('BINARY',dst1)
cv2.imshow('BINARY_INV',dst2)
cv2.imshow('TOZERO',dst3)
cv2.imshow('TOZERO_INV',dst4)
cv2.imshow('TRUNC',dst5)
cv2.waitKey()
cv2.destroyAllWindows()
```

## 自适应阈值

自适应阈值处理能够更好地处理阴暗分布不均的图像，获得更简单的图像效果

`dst = cv2.adaptiveThreshold(src, maxValue, adaptiveMethod, thresholdType, blockSize, C)`\
参数说明:\
src : 被处理图像，必须是灰色图像\
maxValue: 阈值处理采用的最大值\
adaptiveMethod: 自适应阈值的计算方法。 自适应阈值的计算方法
thresholdType: 阈值处理类型; 需要注意的是, 阈值处理类型须是cv2.THRESH_BINARY或者cv2.THRESH_BINARY_INV中的一个\
blockSize: 一个正方形区域的大小。例如,5指的是5X5的区域。\
C: 常量。阈值等于均值或者加权值减去这个常量。\
返回值说明：\
dst : 经过阈值处理后的图像



**自适应阈值的计算方法及其解释**
类型|含义
--|--
cv2.ADAPTIVE_THRESH_MEAN_C|对一个正方形区域内的所有像素平均加权
cv2.ADAPTIVE_THRESH_GAUSSIAN_C|根据高斯函数按照像素与中心点的距离对一个正方形区域内的所有像素进行加权运算

```python
import cv2
image = cv2.imread("../1.2.jpg")
image = cv2.resize(image,(640,320))
image_Gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
athdMEAN = cv2.adaptiveThreshold(image_Gray,255,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,5,3)
athdGAUS = cv2.adaptiveThreshold(image_Gray,255,cv2.cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,5,3)
cv2.imshow("MEAN_C",athdMEAN)
cv2.imshow("GAUSSIAN_C",athdGAUS)
cv2.waitKey()
cv2.destroyAllWindows()
```

## Otsu方法

`retval, dst = cv2.threshold(src,thresh,maxval,type)`
参数说明:\
src : 被处理图像，必须是灰色图像\
thresh： 阈值，且要把阈值设为0\
maxValue: 阈值处理采用的最大值,即255\
type : 阈值处理类型。例如:cv2.THRESH_BINARY + cv2.THRESH_OTSU\
返回值说明：\
dst : 经过阈值处理后的图像\
retval: 由Otsu算法计算后得到的并使用最合适的阈值

```python

import cv2
image = cv2.imread("../1.2.jpg")
image = cv2.resize(image,(640,320))
image_Gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
t1,dst = cv2.threshold(image_Gray,0,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)
cv2.putText(dst,"best threshold"+str(t1),(0,30),cv2.FONT_HERSHEY_SCRIPT_SIMPLEX,1,(255,255,255),2)
cv2.imshow("otsu",dst)
cv2.waitKey()
cv2.destroyAllWindows()
```
