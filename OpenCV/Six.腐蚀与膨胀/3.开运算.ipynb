## 开运算

开运算就是将图像先进行腐蚀操作,再进行膨胀操作。开运算可以用来抹除图像的外部细节(或者噪声)

```python
import cv2
import numpy as np

img = cv2.imread('../spider.jpg')
img = cv2.resize(img,(640,320))
k = np.ones((7,7),np.uint8)
cv2.imshow('img',img)
dst = cv2.erode(img,k)
cv2.imshow('dst1',dst)
k = np.ones((5,5),np.uint8)
dst = cv2.dilate(dst,k)
cv2.imshow('dst2',dst)
cv2.waitKey()
cv2.destroyAllWindows()
```
