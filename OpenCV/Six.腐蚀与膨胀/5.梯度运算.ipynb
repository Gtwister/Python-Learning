## 梯度运算

用原图的膨胀图像减去腐蚀图像,得到轮廓图像。梯度运算的参数为`cv2.MORPH_GRADIENT`

`cv2.morphologyEx(img,cv2.MORPH_GRADIENT, k)`\
参数说明:\
k:是核大小

```python
import cv2
import numpy as np
img = cv2.imread('../spider.jpg')
img = cv2.resize(img,(640,320))
k = np.ones((5,5),np.uint8)
cv2.imshow('img', img)
dst = cv2.morphologyEx(img,cv2.MORPH_GRADIENT,k)
cv2.imshow('dst',dst)
cv2.waitKey()
cv2.destroyAllWindows()
```
