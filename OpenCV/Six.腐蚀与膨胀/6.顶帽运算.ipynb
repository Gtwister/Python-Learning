## 顶帽运算

`dst = cv2.morphologyEx(img, cv2.MORPH_TOPHAT,k)`

```python
import cv2
import numpy as np
img = cv2.imread('../spider.jpg')
img = cv2.resize(img,(640,320))
k = np.ones((7,7),np.uint8)
cv2.imshow('img', img)
dst = cv2.morphologyEx(img,cv2.MORPH_TOPHAT,k)
cv2.imshow('dst',dst)
cv2.waitKey()
cv2.destroyAllWindows()
```
