## 腐蚀

`dst = cv2.erode(src, kernel, anchor, iterations, boderType, boderValue)`\
参数的说明:\
kernel: 腐蚀使用的核\
anchor:  可选的参数,核的锚点位置\
iterations: 可选参数,腐蚀操作的迭代操作,默认值为1.


在opencv做腐蚀或者其它形态学操作时,经常使用Numpy数组来创建核数组,例如:
```python
import cv2
import numpy as np

k= np.one((5,5),np.uint8)
```
行列数越大计算出的效果就越粗糙,行列数越小,计算出的效果就越精细

````python
import cv2
import numpy as np

img = cv2.imread('../spider.jpg')
img = cv2.resize(img,(640,320))
k = np.ones((7,7),np.uint8)
cv2.imshow('img',img)
dst = cv2.erode(img,k)
cv2.imshow('dst',dst)
cv2.waitKey()
cv2.destroyAllWindows()
````
