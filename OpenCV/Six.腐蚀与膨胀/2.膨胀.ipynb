## 膨胀

`dst = cv2.dilate(src, kernel, anchor, iterations, boderType, boderValue)`\
参数说明:\
kernel: 腐蚀使用的核\
anchor:  可选的参数,核的锚点位置\
iterations: 可选参数,腐蚀操作的迭代操作,默认值为1.

```python
import cv2
import numpy as np

img = cv2.imread('../spider.jpg')
img = cv2.resize(img,(640,320))
k = np.ones((3,3),np.uint8)
cv2.imshow('img',img)
dst = cv2.dilate(img,k)
cv2.imshow('dst',dst)
cv2.waitKey()
cv2.destroyAllWindows()
```
