## 图像的轮廓

`contours , hirearchy = cv2.findContours(image, mode, methode)`\
参数说明:\
image: 被检测的图像必须是8位单通道的二值化图像。\
mode: 轮廓的检索模式\
methode: 检测轮廓时使用的方法\
返回值说明:\
contours: 检测出的所有轮廓\
hierarchy: 轮廓之间的层次关系


**轮廓的检索模式的参数值**
参数值|含义
--|--
cv2.RETR_EXTERNAL|只检测外轮廓
cv2.RETR_LIST|检测所有轮廓,但不建立层次关系
cv2.RETR_CCOMP|检测所有轮廓,并建立两级层次关系
cv2.RETR_TREE|检测所有轮廓,并建立树状结构的层次关系

**检测轮廓时的使用方法**
参数值|含义
--|--
cv2.CHAIN_APPROX_NONE|存储轮廓的所有点
cv2.CHAIN_APPROX_SIMPLE|只保存水平、垂直或对角线轮廓的端点
cv2.CHAIN_APPROX_TC89_L1|Ten-Chinl近似算法中的一种
cv2.CHAIN_APPROX_TC89_KCOS|Ten-Chinl近似算法中的一种

`image = cv2.drawContours(image, contours, contourIdx, color, thickness, lineType, hierarchy, maxLevel, offse)`\
参数说明:\
image: 被绘制轮廓的原始图像,可以是多通道图像。\
contours: findContours()方法得出的轮廓列表。\
contourIdx: 绘制轮廓的索引,如果为-1则绘制所有轮廓。\
color: 绘制颜色,使用BGR格式。\
thickness: 可选参数,画笔的粗细程度, 如果该值为-1则绘制实心轮廓\
lineTypee: 可选参数,绘制轮廓的线型。\
hierachy: 可选参数,findContours()方法得出的层次关系。\
maxLevel: 可选参数,绘制轮廓的层次深度,最深绘制在第maxLevel层\
offse: 可选参数,偏移量

```python
import cv2
from matplotlib import pyplot as plt
img = cv2.imread('shape-1.png')
img = cv2.resize(img,(640,320))
gray = cv2.cvtColor(img , cv2.COLOR_BGR2GRAY)
t , binary = cv2.threshold(gray , 127,255,cv2.THRESH_BINARY)
contours , hierarchy = cv2.findContours(binary,cv2.RETR_LIST,cv2.CHAIN_APPROX_NONE)
cv2.drawContours(img,contours,-1,(0,0,255),5)
img2 = img[:,:,::-1]
plt.imshow(img2)
plt.show()
```

**drawContours()方法的第三个参数可以指定绘制哪个索引的轮廓。索引的顺序由轮廓的检索模式决定,例如cv2.RETR_CCOMP模式下绘制索引为0的轮廓**

```python
import cv2
from matplotlib import pyplot as plt
img = cv2.imread('shape-1.png')
img = cv2.resize(img,(640,320))
gray = cv2.cvtColor(img , cv2.COLOR_BGR2GRAY)
t , binary = cv2.threshold(gray , 127,255,cv2.THRESH_BINARY)
contours , hierarchy = cv2.findContours(binary,cv2.RETR_CCOMP,cv2.CHAIN_APPROX_NONE)
cv2.drawContours(img,contours,0,(0,0,255),5) ##索引为0
img2 = img[:,:,::-1]
plt.imshow(img2)
plt.show()
```
