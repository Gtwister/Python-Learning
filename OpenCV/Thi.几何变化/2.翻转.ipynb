## 翻转

`dst = cv2.flip(src, flipCode)`\
参数说明:\
src: 原始图像\
flipCode : 翻转类型

**flip参数值**
参数值|含义
--|--
0|沿着X轴翻转
正数|沿着Y轴翻转
负数|同时沿着X轴、Y轴翻转

```python
import cv2
img =cv2.imread("../1.2.jpg")
img = cv2.resize(img,(320,240))
dst1 = cv2.flip(img,0)
dst2 = cv2.flip(img,1)
dst3 = cv2.flip(img,-1)
cv2.imshow("img",img)
cv2.imshow("dst1",dst1)
cv2.imshow("dst2",dst2)
cv2.imshow("dst3",dst3)
cv2.waitKey()
cv2.destroyAllWindows()
```
