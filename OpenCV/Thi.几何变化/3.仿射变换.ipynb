## 仿射变化

`dst = cv2.warpAffine(src, M, dsize, flags, borderMode, borderValue)`\
参数说明:\
src: 原始图像
M : 一个2行3列的矩阵，根据此矩阵的值变化原图中的像素位置。\
dsize : 输出图像的尺寸大小\
flags : 可选参数, 边界类型, 建议使用默认值。\
borderMode : 可选参数,边界类型,建议使用默认。\
borderValue : 可选参数, 边界值, 默认为0, 建议使用默认值。\

M也被叫作仿射矩阵, 实际上就是2*3的列表,其格式如下:\
`M = [[a, b, c],[d, e, f]]`\
图像做何种仿射变换,完全取决于M的值,输出图像会按以下公式进行计算:\
新x = 原x * a + 原y * b + c\
新y = 原x * d + 原y * e + f\
M矩阵中的数字用32位浮点格式


**平移**

`M = [[1,0,水平移动的距离],[0, 1, 垂直移动的距离]]`


```python
import cv2
import numpy as np
img = cv2.imread('../1.2.jpg')
img = cv2.resize(img,(320,240))
rows = len(img) # 图像像素行数
cols = len(img[0]) # 图像像素列数
M = np.float32([[1,0,50],[0,1,100]])
dst = cv2.warpAffine(img,M,(cols,rows))
cv2.imshow("img",dst)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
import cv2
img = cv2.imread('../1.2.jpg')
img = cv2.resize(img,(320,240))
rows = len(img) # 图像像素行数
cols = len(img[0]) # 图像像素列数
center = (rows/2,cols/2)
M = cv2.getRotationMatrix2D(center, 30 , 0.8)
dst = cv2.warpAffine(img , M, (cols, rows))
cv2.imshow("dst",dst)
cv2.waitKey()
cv2.destroyAllWindows()
```
