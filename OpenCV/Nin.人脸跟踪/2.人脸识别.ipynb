### EigenFaces 人脸识别器

Eigenfaces也被叫做"特征脸",其通过PCA(主成分分析技术)方法将人脸数据转换到另外一个空间维度进行相似度计算。在计算过程中,算法可以忽略掉一些无关紧要的数据,仅识别一些经常有代表性的特征数据,最后根据这些特征数据进行识别。

`recognizer = cv2.face.EigenFaceRecognizer_create(num_componentes, threshold)`


参数说明:
* num_componentes: 可选参数,PCA方法中保留的分量个数,建议使用默认值

* thresh: 可选参数,人脸识别使用的阈值,建议使用默认值

创建完识别器对象后,需要通过对象的train()方法训练识别器,建议每个人至少给出两张以上的照片

`recognizer.train(src, labels)`

参数说明:
* src: 用训练的人脸图像列表,格式为list。样本图像必须宽高一致。

* labels: 样本对应的标签,格式为数组,元素类型为整数。数组长度必须与样本列表长度相同。

训练完识别器之后可以用predict()方法来进行预测

`label,confidence = recognizer.predict(src)`

```python
import cv2
import numpy as np

photos = list()
labels= list()

tao1 = cv2.imread('./faces/tao1.jpg',0)
tao1 = cv2.resize(tao1,(640,320))

tao2 = cv2.imread('./faces/tao2.jpg',0)
tao2 = cv2.resize(tao2,(640,320))

tao3 = cv2.imread('./faces/tao3.jpg',0)
tao3 = cv2.resize(tao3,(640,320))

liang1 = cv2.imread('./faces/liang1.jpg',0)
liang1 = cv2.resize(liang1,(640,320))

liang2 = cv2.imread('./faces/liang2.jpg',0)
liang2 = cv2.resize(liang2,(640,320))

liang3 = cv2.imread('./faces/liang3.jpg',0)
liang3 = cv2.resize(liang3,(640,320))

photos.append(tao1)
labels.append(0)
photos.append(tao2)
labels.append(0)
photos.append(tao3)
labels.append(0)

photos.append(liang1)
labels.append(1)
photos.append(liang2)
labels.append(1)
photos.append(liang3)
labels.append(1)

names = {"0":"Zhangzhentao","1":"Liangshuhang"}

recognizer = cv2.face.EigenFaceRecognizer_create()
recognizer.train(photos,np.array(labels))

i = cv2.imread('./faces/tao4.jpg',0)
i= cv2.resize(i,(640,320))

label, confidence = recognizer.predict(i)
print("confidence = ",str(confidence))
print(names[str(label)])
```

### Fisherfaces 人脸识别器

Fisherfaces 是由Ronald Fisher最早提出的。Fisherfaces通过LDA(线性判别分析技术)方法也会将人脸数据转换到另外一个空间维度去做投影计算,最后根据不同人脸数据的投影距离来判断其相似度。

`recognizer = cv2.face.FisherFaceRecognizer_create(num_components, threshold)`

参数说明:
* num_components: 可选参数,通过Fisherface方法进行判断分析时保留的分量个数,建议使用默认值

* thresh:可选参数,人脸识别使用的阈值,建议使用默认值


`recognizer.train(src, labels)`

`label,confidence = recognizer.predict(src)`

```python
import cv2
import numpy as np

photos = list()
labels= list()

tao1 = cv2.imread('./faces/tao1.jpg',0)
tao1 = cv2.resize(tao1,(640,320))

tao2 = cv2.imread('./faces/tao2.jpg',0)
tao2 = cv2.resize(tao2,(640,320))

tao3 = cv2.imread('./faces/tao3.jpg',0)
tao3 = cv2.resize(tao3,(640,320))

liang1 = cv2.imread('./faces/liang1.jpg',0)
liang1 = cv2.resize(liang1,(640,320))

liang2 = cv2.imread('./faces/liang2.jpg',0)
liang2 = cv2.resize(liang2,(640,320))

liang3 = cv2.imread('./faces/liang3.jpg',0)
liang3 = cv2.resize(liang3,(640,320))

photos.append(tao1)
labels.append(0)
photos.append(tao2)
labels.append(0)
photos.append(tao3)
labels.append(0)

photos.append(liang1)
labels.append(1)
photos.append(liang2)
labels.append(1)
photos.append(liang3)
labels.append(1)

names = {"0":"Zhangzhentao","1":"Liangshuhang"}

recognizer = cv2.face.FisherFaceRecognizer_create()
recognizer.train(photos,np.array(labels))

i = cv2.imread('./faces/tao2.jpg',0)
i= cv2.resize(i,(640,320))

label, confidence = recognizer.predict(i)
print("confidence = ",str(confidence))
print(names[str(label)])
```
