### 级联分类器


将一系列简单的分类器按照一定顺序级联到一起就构成级联分类器,使用级联分类器的程序可以通过一系列简单的判断来对样本进行识别。例如,依次满足"六条腿""有翅膀""有头胸腹"就可以被初步 判断为昆虫。

**加载级联分类器**

`<CascadeClassifier object> = cv2.CascadeClassifier(filename)`

用创建好的分类器对图像进行识别

`objects = cascade.detectMultiScale(image, scaleFacetor, minNeighbors, flags, minSize, maxSize)`

对象说明:\
cascade:已有的分类器对象
参数说明:\
scaleFactor:可选参数,扫描图像时的缩放比例。\
minNeighbors:可选参数，每个候选区域至少保留多少个检测结果才可以判定为人脸。值越大，分析的误差越小。\
minSize: 可选参数,最小的目标尺寸。\
maxSize: 可选参数,最大的目标尺寸。\
返回值说明:\
obj: 目标区域数组,[左上角横坐标,左上角纵坐标,区域宽,区域高]

**分析人脸的位置**

```python
import cv2

img= cv2.imread('./faces.jpg')

facecascade = cv2.CascadeClassifier("./haarcascades/haarcascade_frontalface_default.xml")
faces = facecascade.detectMultiScale(img,1.15)
for (x,y,w,h) in faces:
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,0,255),5)
cv2.imshow('img',img)
cv2.waitKey()
cv2.destroyAllWindows()
```

**标记车牌的位置**

```python
import cv2

img =cv2.imread('./car.jpg')

carcascade = cv2.CascadeClassifier("./haarcascades/haarcascade_russian_plate_number.xml")
cars = carcascade.detectMultiScale(img,1.15,4)

for (x,y,w,h) in cars:
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,0,255),5)
cv2.imshow("img",img)
cv2.waitKey()
cv2.destroyAllWindows()
```
