## 高斯滤波器

`dst = cv2.GaussianBlur(src, ksize, sigmaX, sigmaY, borderType)`\
参数说明:\
sigmaX: 卷积核水平方向的标准差\
sigmaY: 卷积核垂直方向的标准差。修改sigmaX或者sigmaY的值都可以改变卷积核中的权重比例。如果不知道如何设计这两个参数值,就直接把这两个参数的值修改成0,方法会自动计算合适的权重比例.\


```python
import cv2
img = cv2.imread('../1.2.jpg')
img =cv2.resize(img,(320,240))
dst1 = cv2.GaussianBlur(img, (5,5), 0, 0)
dst2 = cv2.GaussianBlur(img, (9,9), 0, 0)
dst3 = cv2.GaussianBlur(img, (15,15), 0, 0)
cv2.imshow('1',img)
cv2.imshow('2',dst1)
cv2.imshow('3',dst2)
cv2.imshow('4',dst3)
cv2.waitKey(0)
cv2.destroyAllWindows()

```
