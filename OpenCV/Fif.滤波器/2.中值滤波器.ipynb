## 中值滤波器

`dst = cv2.medianBlur(src, ksize)`\
参数说明:\
ksize: 滤波核(边长)

```python
import cv2
img = cv2.imread('../1.2.jpg')
img =cv2.resize(img,(320,240))
dst1 = cv2.medianBlur(img, 3)
dst2 = cv2.medianBlur(img, 5)
dst3 = cv2.medianBlur(img, 7)
cv2.imshow('1',img)
cv2.imshow('2',dst1)
cv2.imshow('3',dst2)
cv2.imshow('4',dst3)
cv2.waitKey(0)
cv2.destroyAllWindows()

```
