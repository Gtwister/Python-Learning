```python
import cv2
import time
import numpy as np

width , height = 200,200

r = 20

x = r + 20

y = r + 100

x_offer = y_offer = 4

while cv2.waitKey(1) == -1:
    if x > width - r or x< r:
        x_offer *=-1
    if y > height - r or y < r:
        y_offer *= -1
    x+=x_offer
    y+=y_offer
    img = np.ones((width,height,3),np.uint8)*255
    cv2.circle(img,(x,y),r,(255,0,0),-1)
    cv2.imshow("img",img)
    time.sleep(1/60)

cv2.destroyAllWindows()
```
