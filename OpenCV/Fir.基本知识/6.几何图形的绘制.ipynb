## 线段的绘制

**line()方法**\
`img = cv2.line(img, pt1, pt2, color, thickness)`\
参数说明:\
  pt1:起点坐标\
  pt2:终点坐标\
  color:线条颜色\
  thickness: 绘制线段时的线条宽度  


```python
import cv2
import numpy as np

canvas = np.zeros((300,300,3),np.uint8)
canvas = cv2.line(canvas,(50,100),(150,100),(0,255,255),5)

cv2.imshow("img",canvas)
cv2.waitKey()
cv2.destroyAllWindows()
```

## 绘制一个矩形边框

`img = cv2.rectangle(img, pt1, pt2, color, thickness)`\
参数:\
pt1: 矩形的左上角坐标\
pt2: 矩形的右下角坐标


```python
import cv2
import numpy as np

canvas = np.ones((300,300,3),np.uint8)*255

canvas = cv2.rectangle(canvas,(50,50),(150,200),(255,255,0),2)

cv2.imshow("img",canvas)
cv2.waitKey()
cv2.destroyAllWindows()
```

## 圆形的绘制

`img = cv2.circle(img,center,radius,color,thickness)`\\
参数说明:\\
center: 圆形圆心的坐标\\
radius: 圆形的半径\\
thickness: 绘制圆形时的线条宽度

```python
import numpy as np
import cv2

canvas = np.zeros((300,300,3),np.uint8)

canvas = cv2.circle(canvas, (50, 50),40,(0,0,255), -1)

canvas = cv2.circle(canvas, (150, 50),40,(0,255,255), -1)

canvas = cv2.circle(canvas, (250, 50),40,(0,255,0), -1)

cv2.imshow('img',canvas)
cv2.waitKey()
cv2.destroyAllWindows()
```
