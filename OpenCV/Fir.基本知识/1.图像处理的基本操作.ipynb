## 图像处理的基本操作

**1.1 读取图像**

**`imread()`方法**\
语法: image = cv2.imread(filename,flags)\
参数说明:
* filename: 要读取文件的完整文件名

* flags:读取图像颜色类型标记。当flags的默认值是1时,表示读取的是彩色图像,此时的flags值可以省略;当flags的值为0时,表示读取的是灰色图像.



```python
import cv2

image = cv2.imread("./1.2.jpg")

print(image)
```

**1.2 显示图像**


**`imshow()`方法**\
用途: 用于显示图像\
语法: `cv2.imshow(winname,mat)`\
参数说明：
* winname: 显示图像的窗口名称

* mat: 要显示的图像

**`waitKey()`方法**\
用途:用于等待用户按下键盘上的按键时间\
语法: `retval = cv2.waitKey(delay)`\
参数说明:
* retval:与被按下的按键相对应的ASCLL码\
* delay:等待用户按下键盘上按键的时间,单位为毫秒(ms)。当delay的值为负数、0或者为空时,表示无限等待用户按下键盘的时间。

**`destroyAllWindows()`方法**\
用途:用于销毁所有正在显示图像的窗口


```python
import cv2
image = cv2.imread("./1.2.jpg")

cv2.imshow("1",image)

cv2.waitKey()

cv2.destroyAllWindows()
```

**保存图像**

语法: `cv2.imwrite(filename,img)`\
用途: 按照指定路径保存图像的imwrite()方法
* filename: 保存图像时所用的完整路径

* img: 要保存的图像

```python
import cv2

image = cv2.imread('./1.2.jpg')
# 把1.2.jpg保存在Picture文件夹
cv2.imwrite('./Picture/2.jpg',image)
```

**获取图像属性**

* shape: 如果是彩色图像,那么获取的是一个由图像的像素列数、像素行数和通道数所组成的数组,如果是灰度图像,那么获取的是一个包含图像的像素列数、像素行数的数组,即(像素行数,像素列数)

* size: 获取的是图像包含的像素个数,其值为"像素列数 * 像素行数 * 通道数"。(灰度图像的通道数为1)

* dtype: 获取的是图像的数据类型。

```python
import cv2

image_Color = cv2.imread('./1.2.jpg')
print('获取彩色图像的属性:')
print('shape=',image_Color.shape)
print('size=',image_Color.size)
print('dtype=',image_Color.dtype)
image_Gray = cv2.imread('./1.1.jpg',0)
print('获取灰度图像的属性：')
print('shape=',image_Gray.shape)
print('size=',image_Gray.size)
print('dtype=',image_Gray.dtype)
```
