**绘制文字"mrsoft"**

putText( )方法\
`img = cv2.putText(img, text, org, fontFace, fontScale, color, thickness, lineType, bottomLeftOrigin)`\
参数说明:\
text: 要绘制的文字内容\
org: 文字在画布左下角的坐标\
fontFace: 字体样式，可选参数\
fontScale: 字体大小\
lineType: 线型(指的是线产生的算法，有4和8两个值，默认值是8)\
bottomLeftOrigin: 绘制文字时的方向(有True和False)

```python
import numpy as np
import cv2

canvas = np.zeros((100,300,3),np.uint8)

cv2.putText(canvas,"mrsort",(20,70),cv2.FONT_HERSHEY_TRIPLEX,2,(0,255,0), 5)

cv2.imshow("Text",canvas)
cv2.waitKey()
cv2.destroyAllWindows()
```
