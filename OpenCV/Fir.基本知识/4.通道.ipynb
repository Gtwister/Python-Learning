## 1、拆分通道

**拆分BGR通道**

`b,g,r = cv2.split(bgr_image)`(b,g,r)\
参数:\
b : B通道图像\
g : G通道图像\
r : R通道图像\
bgr_image: 一副BGR图像


```python
import cv2
```

```python
bgr_image = cv2.imread("1.2.jpg")
cv2.imshow('1.2',bgr_image)
b,g,r=cv2.split(bgr_image)
cv2.imshow('B Channel',b)
cv2.imshow('G Channel',g)
cv2.imshow('R Channel',r)
cv2.waitKey()
cv2.destroyAllWindows()
```

**拆分HSV图像**


```python
bgr_image = cv2.imread("1.2.jpg")
hsv_image = cv2.cvtColor(bgr_image,cv2.COLOR_BGR2HSV)
cv2.imshow('1.2',hsv_image)
h,s,v=cv2.split(hsv_image)
cv2.imshow('H Channel',h)
cv2.imshow('S Channel',s)
cv2.imshow('V Channel',v)
cv2.waitKey()
cv2.destroyAllWindows()
```

## 2、合并通道

`bgr = cv2.merge([b,g,r])`\
参数:
[b,g,r]将<u>拆分后</u>的通道顺序合并

```python
image = cv2.imread("1.2.jpg")
b,g,r=cv2.split(image)
bgr = cv2.merge([b,g,r])
cv2.imshow("BGR",bgr)
rgb = cv2.merge([r,g,b])
cv2.imshow("RGB",rgb)
cv2.waitKey()
cv2.destroyAllWindows()
```
