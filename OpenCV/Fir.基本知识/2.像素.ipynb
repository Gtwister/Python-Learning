## 像素

**获取像素的BGR值**

```python
import cv2

image = cv2.imread('./1.2.jpg')

px = image[799,1279]

print("坐标(1279,799)上的像素的BGR值是",px)
```

**修改指定区域内的所有像素**

```python
import cv2

image = cv2.imread("1.2.jpg")
for i in range(1000,1051):
    for j in range(500,551):
        image[j,i] = [255,255,255]
cv2.imshow("2.8",image)
cv2.waitKey()
cv2.destroyAllWindows()
```
