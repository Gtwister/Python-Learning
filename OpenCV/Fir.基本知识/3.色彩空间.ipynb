## 1、GRAY色彩空间

**从RGB/BGR空间转化成GRAY空间**

`cv2.cvtColor(src,code)`\
参数:\
scr: 转化前的初始图像\
code: 色彩空间转换码

色彩空间转换码|含义
--|--
`cv2.COLOR_BGR2GRAY`|BGR转化为GRAY
`cv2.COLOR_RGB2GRAY`|RGB转化为GRAY

```python
import cv2
```

```python
image = cv2.imread("1.2.jpg")
cv2.imshow("2.1",image)

gray_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)

cv2.imshow("gray",gray_image)

cv2.waitKey()
cv2.destroyAllWindows()
```

## 2、HSV色彩空间

HSV色彩空间是基于色调(H),饱和度(S),亮度(V)

**从RGB/BGR空间转化成HSV空间**

色彩空间转换码|含义
--|--
`cv2.COLOR_BGR2HSV`|BGR转化为HSV
`cv2.COLOR_RGB2HSV`|RGB转化为HSV

```python
image = cv2.imread("1.2.jpg")
cv2.imshow("2.1",image)

gray_image = cv2.cvtColor(image,cv2.COLOR_BGR2HSV)

cv2.imshow("gray",gray_image)

cv2.waitKey()
cv2.destroyAllWindows()
```
