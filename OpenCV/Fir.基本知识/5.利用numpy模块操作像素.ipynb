## Numpy模块操作像素

**创建黑色图像**

```python
import cv2
import numpy as np
```

```python
width = 200
height = 100
img = np.zeros((height,width),np.uint8)
cv2.imshow("img",img)
cv2.waitKey()
cv2.destroyAllWindows()
```

**创建纯白色图像**

```python
width = 200
height = 100
img = np.ones((height,width),np.uint8)*255
cv2.imshow("img",img)
cv2.waitKey()
cv2.destroyAllWindows()
```

**创建黑白相间的图案**

```python
width = 200
height = 100
img = np.zeros((height,width),np.uint8)
for i in range(0,width,40):
    img[:,i:i+20]=225
cv2.imshow("img",img)
cv2.waitKey()
cv2.destroyAllWindows()
```

**创建彩色图像**

```python
width = 200
height = 100
img = np.zeros((height,width,3),np.uint8)
blue = img.copy()
blue[:,:,0]=255
green = img.copy()
green[:,:,1]=255
red = img.copy()
red[:,:,2]=255
cv2.imshow('red',red)
cv2.imshow('green',green)
cv2.imshow("blue",blue)
cv2.waitKey()
cv2.destroyAllWindows()
```

## 创建随机图像

```python
import cv2
import numpy as np

width  = 200
height = 100
img = np.random.randint(256,size=(height,width,3),dtype = np.uint8)# 创建指定宽高、单通道、随机像素值的图像，随机值在0~256之间数字为无符号整数8位整数
cv2.imshow("img",img)
cv2.waitKey()
cv2.destroyAllWindows()
```

## 水平或者垂直拼接图像

**Numpy库中给予了hstack和vstack函数，同样，可以根据图像像素来进行拼接** 

```python
import cv2
import numpy as np
img1 = cv2.imread("./1.2.jpg")

img_h = np.hstack((img1,img1)) # 竖直拼接
img_w = np.vstack((img1,img1)) # 水平拼接

cv2.imshow("img_h",img_h)
cv2.imshow("img_w",img_w)

cv2.waitKey()
cv2.destroyAllWindows()
```
