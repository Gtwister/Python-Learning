## 字符串函数

add( )\
对两个数组元素进行字符串连接

```python
import numpy as np
print(np.char.add(['sun'],['light']))
print(np.char.add(['sun','woc'],['light','nb']))
```

multiply( )\
返回按数组元素多重连接后的字符串

```python
print(np.char.multiply('good',3))
```

center()\
将字符串居中，并使用指定字符的左侧和右侧进行填充

```python
print(np.char.center(["nice","handsome"],20,fillchar="*"))##以20个字符,填充“*”
```

capitalize()\
将字符串第一个字母转换成大写

```python
print(np.char.capitalize(["sun is a bitch","kai Ge is A pieCe of shit"]))
```

title()\
将字符串的每一个单词的第一个字母转换成大写

```python
print(np.char.title(["sun is a bitch","kai Ge is A pieCe of shit"]))
```

lower()\
数组元素转换成小写

```python
print(np.char.lower(["sUn is A biTch","kai Ge is A pieCe of shit"]))
```

upper()\
数组元素转换为大写

```python
print(np.char.upper(["sUn is A biTch","kai Ge is A pieCe of shit"]))
```

split()\
指定分割符对字符串进行分割，并返回数组列表

```python
print(np.char.split(["sUn is A biTch","kai Ge is A pieCe of shit"]))
```

splitlines()\
返回元素中的行列表，以换行符分割

```python
print(np.char.splitlines(["sU\n is A biTch","kai Ge is A p\nieCe of shit"]))
```

strip()\
移除元素开头或者结尾处的特定字符

```python
print(np.char.strip(["*****is A biTch***","*****kai Ge is A pieCe of shit****"],"*"))
```

join()\
通过指定分割符来连接数组中的元素

```python
print(np.char.join("-",["nima","sile"]))
print(np.char.join(["-",":"],["nima","sile"]))
```

replace()\
使用新的字符串来替换字符串中的子字符串

```python
print(np.char.replace(["hai li is a good man","hai li is not a good man"],"good","handsome"))
```

encode()\
编码,数组元素依次调用str.encode()

```python
print(np.char.encode(['sun','light'],"utf-8"))
```

decode()\
解码,数组元素依次调用str.decode()

```python
a = np.char.encode(['sun','light'],"utf-8")
print(np.char.decode(a,"utf-8"))
```
