## 翻转数组


transpose() \
原型：`numpy.transpose(a,axes=None)` \
作用：对换数组维度 \
注意：会修改原始数组 \
参数|说明
--|--
a|要操作的元素
axes|整数列表，对应维度，通常所有维度都会对换


```python
import numpy as np
a = np.arange(12).reshape(3,4)
print(a)
print("\n")

b = np.transpose(a)
print(b)
```

```python
a = np.arange(12).reshape(3,4)
print(a)
print("\n")

b=a.T
print(b)
```

rollaxis() \
原型：`numpy.rollaxis(a,aixs,start=0)`\
作用：向后滚动到一个特定的位置 \
参数|说明
--|--
a|要操作的元素
aixs|向后滚动的轴，其它轴不会改变
start| 默认为0，表示完整的滚动，会滚动到特定的位置

```python
a = np.arange(12).reshape(3,4)
print(a)
print(a.shape)
print("\n")

b = np.rollaxis(a,1)#只有0维度和1维度
print(b)
print("\n")
```

```python
a = np.arange(12).reshape(3,2,2)
print(a)
print(a.shape)
print("\n")

b = np.rollaxis(a,2)
print(b)
print(b.shape)
print("\n")
```

```python
a = np.arange(48).reshape(2,2,3,4)
print(a)
print(a.shape)
print("\n")

b = np.rollaxis(a,3,start=1)
print(b)
print(b.shape)
print("\n")
```

swapaxes() \
原型：`numpy.swapaxes(a,axis1,axis2)` \
作用：交换数组的两个轴 
参数|说明
--|--
a|要操作的元素
aixs1|对应第一个轴的整数
start|对应第二个轴的整数

```python
a = np.arange(24).reshape(2,3,4)
print(a)
print(a.shape)
print("\n")

b = np.swapaxes(a,2,0)
print(b)
print(b.shape)
```
