```python
#多维数组仍然可以用上述方法
arr =np.arange(15)
arr.shape=(5,3)
print(arr)
arr2=arr[2][1]
print(arr2,type(arr2))
```

```python
arr =np.arange(15)
arr.shape=(5,3)
print(arr)
print("\n")
arr2=arr[2:]
print(arr2,type(arr2))
print(arr2.shape)
```

```python
#...取值 表示所有
import numpy as np
arr =np.arange(15)
arr.shape=(5,3)
print(arr)
print("\n")
print(arr[...,1])#取一列
print("\n")
print(arr[1,...])#取一行
print("\n")
print(arr[1:3,...])#取从1到3行，但不包括第3行
print("\n")
print(arr[...,1:])#取从1列到所有列
```
