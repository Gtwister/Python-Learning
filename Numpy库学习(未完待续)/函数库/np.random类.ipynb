np.random.seed()\
原型:`numpy.random.seed(num)`\
作用:生成一个list顺序固定，但是内容未知

```python
import numpy as np

np.random.seed(0)
print(np.random.rand(5))#随机生成五个数
print(np.random.rand(5))
np.random.seed(0)
for i in range(10):
    print(np.random.random())##生成的数据与上面完全一致，说明生成的list相同
np.random.seed(5)
print(np.random.rand(5))
```

np.random.normal()\
原型:`numpy.random.normal(loc,scale,size)`\
作用:生成一个满足参数正太分布的数组
参数|说明
 --|--
 loc|正态分布的均值
 scale|正太分布的标准差
 size|设定数组的形状

```python
import numpy as np
a = np.random.normal(loc=0,scale=2,size=(2,3))
a
```
