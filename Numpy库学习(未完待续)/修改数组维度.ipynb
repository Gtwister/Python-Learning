## 修改数组维度

broadcast() \
用于模仿广播的对象，它返回一个对象，该对象封装了将一个数组广播到另一个数组的结果

```python
import numpy as np

x = np.array([[1],[2],[4]])
y = np.array([4,5,6])
print(x)
print(y)
```

```python
b = np.broadcast(x,y)##b是一个对象
print(b)
```

```python
r,c = b.iters
print(r)##r迭代对象
print(c)##c迭代对象
```

```python
for a in range(0,6):
    print(next(r),next(c))
```

```python
b = np.broadcast(x,y)
print(b.shape)
```

```python
c = np.empty(b.shape)
print(c)
```

```python
print(x+y)
```

broadcast_to() \
原型:`numpy.broadcast_to(array,shape,subok=False)` \
作用:将数组广播到新形状。它在原始数组上返回只读视图。它通常不连续。如果新形状不符合Numpy的广播规则，该函数可能会抛出ValueError
 参数|说明
 --|--
 array|待修改数组
 shape|修改后的形状

```python
import numpy as np
a = np.arange(1,5).reshape(1,4)
print(a)
b = np.broadcast_to(a,(3,4))##不能将列修改得比4小或者比4大，不符合广播规则
print(b)
```

expand_dims() \
原型:`numpy.squeeze(arr,axis=None)`\
作用:通过在指定位置插入新的轴来扩展数组形状
 参数|说明
 --|--
 arr|待修改数组
 axis|新轴插入的位置

```python
x = np.arange(1,5).reshape((2,2))
print(x)
print(x.shape,x.ndim)
```

```python
y = np.expand_dims(x,axis=0)##增加了一个轴，一个维度
print(y)
print(y.shape,y.ndim)
```

squeeze() \
原型:`numpy.squeeze(arr,axis)`\
作用:从给定数组形状中删除一维的条目
 参数|说明
 --|--
 arr|输入数组
 axis|删除轴的位置

```python
import numpy as np
x = np.arange(1,10).reshape((1,3,3))
print(x)
print(x.shape,x.ndim)
```

```python
y = np.squeeze(x)
print(y)
print(y.shape,y.ndim)
```

```python
y = np.squeeze(x)
y = np.squeeze(y)##不会在继续裁剪轴
print(y)
print(y.shape,y.ndim)
```

```python
x = np.arange(1,121).reshape(2,3,4,5)
y = np.squeeze(x)
print(y)
print(y.shape,y.ndim)
```

```python
x = np.arange(1,13).reshape(1,1,3,4)
y = np.squeeze(x)##不加axis参数则会将所有维度为1的轴删除
print(y)
print(y.shape,y.ndim)
```

```python
x = np.arange(1,13).reshape(1,1,3,4)
y = np.squeeze(x,axis=1)##不加axis参数则会将所有维度为1的轴删除
print(y)
print(y.shape,y.ndim)
```
