## 搜索函数

max(),min()\
作用:沿给定轴返回最大和最小值

```python
import numpy as np
x = np.array([[30,40,70],[80,30,10],[50,90,60]])

print(x)
```

```python
print(x.flatten())##数组展开
```

```python
print(np.max(x,axis=0))
print(np.max(x,axis=1))
```

argmax(),argmin()\
作用：返回最大最小值的索引

```python
print(np.argmax(x))
```

```python
print(np.argmax(x,axis=0))
print(np.argmax(x,axis=1))
```

nonzero()\
作用：返回数组中非零数组的索引

```python
x = np.array([[0,40,70],[80,30,10],[50,90,60]])

print(x)
```

```python
print(np.nonzero(x))##返回两组索引,分别对应行和列
```

where()\
作用:返回数组输入数组中满足给定元素的索引

```python
x = np.array([[0,40,70],[80,20,10],[50,90,60]])
print(x)
```

```python
y = np.where(x>30)
print(y)
print('\n')
```

```python
print(x[y])
```

extract()\
作用:根据某个条件从数组中抽取元素，返回满条件的元素
