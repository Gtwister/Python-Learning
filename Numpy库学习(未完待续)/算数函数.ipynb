## 算数函数

add(),subtract(),multiply(),divide()\
作用:加减乘除

```python
import numpy as np

x = np.arange(9,dtype='f').reshape(3,3)
print("x数组：")
print(x)
print("\n")

y = np.array([10,10,10])
print("y数组：")
print(y)
print("\n")

print(np.add(x,y))
print("--------------------")
print(x+y)
```

reciprocal()\
作用:返回参数逐个元素的倒数

```python
x = np.array([0.25,0.5,1,100,0.333])
print(np.reciprocal(x))
```

power()\
作用:将第一个输入数组中的元素作为底数，计算它与第二个输入数组中相应元素的幂

```python
x = np.array([2,3,4])
y = np.array([1,2,3])
print(np.power(x,y))
```

mod(),remainder()\
作用:计算输入数组中相应元素的相除后的余数

```python
x = np.array([10,10,10])
y = np.array([2,3,4])
print(np.mod(x,y))
print(np.remainder(x,y))
```
