```python
#结构化数据类型
import numpy as np
student=np.dtype([("name","S20"),("age","i4"),("mark","f4")])## "S20"表示字符串，string of 20 最长20个字符；"i4" int of 4 bytes 表示4字节整数；"f4"表示4字节浮点数
arr = np.array([("sunck",18,99.999),("kaige",19,98.99)],dtype=student)
arr
```

```python
#arange函数 根据start与stop指定的范围以及step设定的步长，生成一个ndarray数组
arr = np.arange(1,11,1)#(start,stop(拿不到),step)
arr
```

```python
#frombuffer函数  用于实现动态数组，接受buffer输入参数，以流的形式读入转化成ndarray对象
import numpy as np
x=b'show show way'#attention:buffer是字符串时，py3默认str是Unicode类型，所以要转成bytestring在原str前加上b
arr = np.frombuffer(x,dtype='S1',count=5,offset=2)
arr
```

```python
x=[1,2,3,4,5]
z = iter(x)#将x变成可迭代对象
print(z)
print (type(z))
arr = np.fromiter(z,dtype='f')
arr
```

```python
#linspace函数 创建一个等差数列
arr = np.linspace(1,10,num=5,endpoint=True,dtype=int)# endpoint是否包含最后一个数
arr
```

```python
#logspace函数 创建一个等比数列
arr = np.logspace(1,10,num=10,endpoint=True,dtype=int,base=2)
arr
```
