## a.T矩阵转置 遍历顺序
如果对转置改变它的数据存储方式，那么就会改变它的遍历顺序

```python
import numpy as np 
a = np.arange(12)
a = a.reshape(3,4)
print(a,end="\n")
for x in np.nditer(a):#nditer迭代器
    print(x,end=" ,")
print(" ")

print(a.T)
print("\n")
for x in np.nditer(a.T):
    print(x,end=",")
print("\n")

b = a.T.copy(order="C")##这个新的数组存储的数据格式就是按转置之后的数据格式 column
print(b)
print("\n")
for x in np.nditer(b):
    print(x,end=",")
print("\n")
```

## 控制遍历的顺序
改变参数order="C"进行 行优先存储
       order="F"     列

```python
c = a.T.copy(order="C")
for x in np.nditer(c):
    print(x,end=",")
print("\n")
```

```python
c = a.T.copy(order="F")
for x in np.nditer(c):
    print(x,end=",")
print("\n")
```

```python
import numpy as np 
a = np.arange(12)
a = a.reshape(3,4)
print(a,end="\n")
for x in np.nditer(a,order="F",op_flags=["readwrite"]):#op_flags修改只读参数，设为可以写入
    x[...]=x**2 ##对 x 进行“原地（in-place）修改”，将 x 中的每一个元素乘以 2，等价于x[:] = x*2
print(" ")
print(a)
```

```python
import numpy as np 
a = np.arange(12)
a = a.reshape(3,4)
print(a,end="\n")
for x in np.nditer(a,flags=["external_loop"],order="F"):#flags将一个零维数组以一维数组的形式输出
    print(x,end=",")
print(" ")
```

## 广播遍历

```python
a = np.arange(12).reshape(3,4)
a
```

```python
b = np.arange(1,5)
b
```

```python
for x, y in np.nditer([a,b]):##必须可以进行广播
    print("%d:%d"%(x,y),end=",")
print("")
```
