## 数学函数

```python
import numpy as np

a = np.array([0,30,60,90])

sin = np.sin(a*np.pi/180)
cos = np.cos(a*np.pi/180)
tan = np.tan(a*np.pi/180)
print(sin)
print(cos)
print(tan)
```

around( )\
原型:`numpy.around(a,decimals=0)`\
作用:返回指定数字的四舍五入值
参数|说明
 --|--
 a|数组
 decimals|舍入的小数位数。默认为0。如果为负，整数将四舍五入到小数点左侧的位置

```python
x = np.array([3.125,456.325,5471.12,520])
print(np.around(x))
print(np.around(x,decimals=1))
print(np.around(x,decimals=-1))
```
