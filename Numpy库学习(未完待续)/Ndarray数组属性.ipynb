```python
#ndim秩，即是轴的数量或维度的数量
import numpy as np
arr = np.arange(24)
print (arr)
print(arr.ndim)
arr2=arr.reshape(2,3,4)
print(arr2)
print(arr2.ndim)
```

```python
#shape表示维度数，即是行数和列数
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
print(arr.shape)
```

```python
arr = np.array([[1,2,3],[4,5,6]])
arr.shape=(3,2)
print(arr)
print(arr.shape)
```

```python
#size数组元素数量  dtype类型
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
print(arr.size)
print(arr.dtype)
```

```python
#itemsize对象中每个元素的大小，以字节为单位
arr1 = np.array([1,2,3,4,5],dtype='i1')
print(arr1)
print(arr1.itemsize)

arr2 = np.array([1,2,3,4,5],dtype='f')
print(arr2)
print(arr2.itemsize)
```

```python
#flags对象的内存信息
arr = np.array([1,2,3,4])
print(arr.flags)
```
