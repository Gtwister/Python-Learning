## 类与对象

类和对象的关系：用类去创建一个对象。

### 2.1.1 类

类是用来创建对象的，是一系列相同特征和行为的统称。
* 特征即是属性
* 行为即是方法


### 2.1.2 对象

开发中先有类再有对象

### 2.2.1 定义类

* 语法
```python

     class 类名():
          代码
          ....
```

### 2.2.2 创建对象

* 语法
```python

     对象名 =  类名
```

**体验面对对象**

````python
import numpy

class Washer():
    def wash(self):
        print('wash your clothes!')

hair = Washer()
hair.wash()
````

### 2.2.3 类里的self

self 是指调用类里的对象

````python
import numpy

class Washer():
    def wash(self):
        print('wash your clothes!')
        print(self) ##查看self地址和类型

hair = Washer()
print('hair:',hair)
hair.wash()
````
