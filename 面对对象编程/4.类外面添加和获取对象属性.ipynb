### 类外面获取和添加对象属性

属性即是特征，对象属性可以在类里添加或者获取，也可以在类外面添加和获取。

```python
class Washer():
    def wash(self):
        print('wash your clothes!')

haier1 = Washer()

# 对象名.属性名 = 值
haier1.width = 500
haier1.height = 800
# 获取属性 对象名.属性名
print(f'洗衣机的宽度是{haier1.width}')
print(f'洗衣机的高度是{haier1.height}')
```

### 类里面获取对象属性

`self.属性名`

```python
class Washer():
    def print_info(self):
        print(f'洗衣机的宽度是{self.width}')
        print(f'洗衣机的高度是{self.height}')

haier1 = Washer()

# 对象名.属性名 = 值
haier1.width = 500
haier1.height = 800


haier1.print_info()
```
