## 多进程完成多任务

**3.1 进程的创建步骤**

* 1. 导入进程包\
`import multiprocessing as mp`
* 2. 通过进程类创建进程对象\
`progress = mp.Process()`
* 3. 启动进程执行任务\
`progress.start()`

```python
import multiprocessing as mp
```

**3.2 通过进程类创建进程对象**\
`进程对象 = mulyiprogressing.Progress(target=任务名)`
参数名|说明
--|--
target|执行的目标任务名,这里指的是函数名(方法名)
name|进程名,一般不用设置
group|进程组,目前只能使用None

**3.3 进程创建与启动代码**

```python
# 创建子进程
sing_process = mp.Process(target=sing)
# 创建子进程
dance_process = mp.Process(target=dance)
# 启动进程
sing_process.start()
dance_process.start()
```

## **实例**

**单任务**

```python
import time


# 唱歌
def sing():
    for i in range(3):
        print('唱歌...')
        time.sleep(0.5)

# 跳舞
def dance():
    for i in range(3):
        print('跳舞...')
        time.sleep(0.5)

if __name__=='__main__':
    sing()
    dance()
```

```python
import multiprocessing as mp
import time


# 唱歌
def sing():
    for i in range(3):
        print('唱歌...')
        time.sleep(0.5)

# 跳舞
def dance():
    for i in range(3):
        print('跳舞...')
        time.sleep(0.5)

if __name__=='__main__':
    sing_process = mp.Process(target=sing)
    dance_process = mp.Process(target=dance)

    sing_process.start()
    dance_process.start()
```
