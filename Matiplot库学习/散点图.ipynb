# 散点图
### 概念:
1.散点图显示两组数据的值，每个点的坐标位置由变量的值决定。\
2.由一组不连接的点完成，用于观察两种变量的相关性。\
3.例如身高-体重、温度-纬度、等等

```python
import numpy as np
import matplotlib.pyplot as plt
height = [161,170,182,175,173,165]
weight = [50,58,80,70,69,55]

plt.figure()
plt.scatter(height,weight)
plt.show()
##不相关
N=1000
x = np.random.randn(N)
y1 = np.random.randn(N)

plt.figure()
plt.scatter(x,y1)
plt.show()
##正相关
N=1000
x = np.random.randn(N)
y1 = x+np.random.randn(N)*0.5

plt.figure()
plt.scatter(x,y1)
plt.show()
```

scatter()
参数|说明
--|--
c|颜色
s|点大小
alpha|透明度
marker|点形状

```python
N=1000
x = np.random.randn(N)
y1 = x+np.random.randn(N)*0.5

plt.figure()
plt.scatter(x,y1,c='red',s=100,marker='<',alpha=0.1)
plt.show()
```
