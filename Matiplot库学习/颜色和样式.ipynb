## 颜色
1.默认八种颜色缩写：b:blue  g:green  r:red c:cyan m:magenta  y:yellow  k:black  w:white\
2.其它颜色表示方法:\
灰色阴影\
html 十六进制\
RGB元组

```python
import numpy as np
import matplotlib.pyplot as plt

y = np.arange(1,5)

plt.plot(y,color='c')

plt.plot(y+1,color='0.5')##灰度值

plt.plot(y+2,color='#FF00FF')##十六进制

plt.plot(y+3,color=(0.1,0.2,0.3))##RGB颜色通道
plt.show()
```

## 点线的样式
23种点形状:\


```python
import numpy as np
import matplotlib.pyplot as plt

y = np.arange(1,5)

plt.plot(y,marker='o')##显式指定marker则会画出线段

plt.plot(y+1,marker='D')##参数唯一，可识别不用写参数名称

plt.plot(y+2,'x')

plt.plot(y+3,'p')
plt.show()
```

四种线式:

'  -  ':实线\
' -- ':虚线\
' -. ':点划线\
' : ':点线\

```python
import numpy as np
import matplotlib.pyplot as plt

y = np.arange(1,5)

plt.plot(y,'--')

plt.plot(y+1,'.-')

plt.plot(y+2,':')

plt.plot(y+3,'-')
plt.show()
```
