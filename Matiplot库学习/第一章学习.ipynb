## 2.1 基本用法

```python
##%matplotlib 魔法指令可以让jupyter book出现图像窗口
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(-1,1,50)##linspace划分区间
y=x**2
plt.plot(x,y)##plot(x(横坐标),y(纵坐标))位置不要错位
plt.show()##show()展示图像
```

## 2.2 figure图像

```python
x = np.linspace(-3,3,50)##linspace划分区间
y1=x**2
y2=2*x+1
plt.figure()##分开图形窗口
plt.plot(x,y1)
plt.figure(num=3,figsize=(8,5))##num=(设置图片序号),figsize=(设置画布大小)
plt.plot(x,y2)
plt.plot(x,y1,color='red',linewidth=1.0,linestyle='--')##color=线的颜色,linewidtg=线的宽度,linestyle=线的格式
plt.show()
```

## 2.3 设置坐标轴1

```python
x = np.linspace(-3,3,50)##linspace划分区间
y1=x**2
y2=2*x+1

plt.figure(num=3,figsize=(8,5))##num=(设置图片序号),figsize=(设置画布大小)
plt.plot(x,y2)
plt.plot(x,y1,color='red',linewidth=1.0,linestyle='--')##color=线的颜色,linewidtg=线的宽度,linestyle=线的格式
plt.xlim((-1,2))##限制x轴的取值范围
plt.ylim((-2,3))##限制y轴的取值范围
plt.xlabel('I am X')##描述x轴名称
plt.ylabel('I am Y')

new_ticks= np.linspace(-1,2,5)
##print(new_ticks)
plt.xticks(new_ticks)##更换坐标名称
plt.yticks([-2,-1.8,-1,1.22,3],[r'$really\ bad$',r'$bad\ \alpha$',r'$normal$',r'$good$',r'$really\ good$'])
##分别对应尺度，用r'$words$'来表示正则表达，使得字体更加美观，但是$符不能读取空格所以要用\代替,\alpha表示数学形式的
plt.show()
```

## 2.4 设置坐标轴2

```python
x = np.linspace(-3,3,50)##linspace划分区间
y1=x**2
y2=2*x+1

plt.figure(num=3,figsize=(8,5))##num=(设置图片序号),figsize=(设置画布大小)
plt.plot(x,y2)
plt.plot(x,y1,color='red',linewidth=1.0,linestyle='--')##color=线的颜色,linewidtg=线的宽度,linestyle=线的格式
plt.xlim((-1,2))##限制x轴的取值范围
plt.ylim((-2,3))##限制y轴的取值范围
plt.xlabel('I am X')##描述x轴名称
plt.ylabel('I am Y')

new_ticks= np.linspace(-1,2,5)
##print(new_ticks)
plt.xticks(new_ticks)##更换坐标名称
plt.yticks([-2,-1.8,-1,1.22,3],[r'$really\ bad$',r'$bad\ \alpha$',r'$normal$',r'$good$',r'$really\ good$'])
##分别对应尺度，用r'$words$'来表示正则表达，使得字体更加美观，但是$符不能读取空格所以要用\代替,\alpha表示数学形式的

ax = plt.gca()##gca='get current axis'
ax.spines['right'].set_color('none')##将右边坐标轴变为透明
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')
ax.spines['bottom'].set_position(('data',0))##以data做参照设置轴的位置
ax.spines['left'].set_position(('data',0))
plt.show()
```

## 2.5 Legend 图例

```python
x = np.linspace(-3,3,50)##linspace划分区间
y1=x**2
y2=2*x+1

plt.figure(num=3,figsize=(8,5))##num=(设置图片序号),figsize=(设置画布大小)
plt.plot(x,y2)
plt.plot(x,y1,color='red',linewidth=1.0,linestyle='--')##color=线的颜色,linewidtg=线的宽度,linestyle=线的格式
plt.xlim((-1,2))##限制x轴的取值范围
plt.ylim((-2,3))##限制y轴的取值范围
plt.xlabel('I am X')##描述x轴名称
plt.ylabel('I am Y')

new_ticks= np.linspace(-1,2,5)
##print(new_ticks)
plt.xticks(new_ticks)##更换坐标名称
plt.yticks([-2,-1.8,-1,1.22,3],[r'$really\ bad$',r'$bad\ \alpha$',r'$normal$',r'$good$',r'$really\ good$'])
##分别对应尺度，用r'$words$'来表示正则表达，使得字体更加美观，但是$符不能读取空格所以要用\代替,\alpha表示数学形式的
l1,=plt.plot(x,y2,color='blue',label='up')##label=设置线的标签,'l1,'为了传入handles参数中所用的实例
l2,=plt.plot(x,y1,color='red',linewidth=1.0,linestyle='--',label='down')
plt.legend(handles=[l1,l2,],labels=['aaa','bbb'],loc='best')##画出legend图示，表示线的对应标签,handles=[传入需要的线],labels=[更改后的标签],loc=设置legend图示位置
plt.show()
```

## 2.6 Annotations 标注

```python
x = np.linspace(-3,3,50)##linspace划分区间
y1=x**2
y2=2*x+1

plt.figure(num=3,figsize=(8,5))##num=(设置图片序号),figsize=(设置画布大小)
plt.plot(x,y2)

ax = plt.gca()##gca='get current axis'
ax.spines['right'].set_color('none')##将右边坐标轴变为透明
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')
ax.spines['bottom'].set_position(('data',0))##以data做参照设置轴的位置
ax.spines['left'].set_position(('data',0))

x0 = 1
y0 = 2*x0+1
plt.scatter(x0,y0,s=50,color='r')##scatter()画点
plt.plot([x0,x0],[y0,0],'k--',lw=2.5)##plot([x1,x2],[y1,y2])以(x1,y1)和(x2,y2)两点连一条直线,'k--'=简写以虚线的格式,lw=linewidth
#method 1
##############
plt.annotate(r'$2x+1=%s$'%y0,xy=(x0,y0),xycoords='data',xytext=(+30,-30),textcoords='offset points',fontsize=16,arrowprops=dict(arrowstyle='->',connectionstyle='arc3,rad=.2'))

#method 2
##############
plt.text(-3.7,3,r'$This\ is\ the \ text.\ \mu\ \sigma_i\ \alpha_t$',fontdict={'size':16,'color':'r'})

plt.show()
```

```python
x = np.linspace(-3,3,50)
y=0.1*x

plt.figure()
plt.plot(x,y,linewidth=10,zorder=1)
plt.ylim(-2,2)
ax = plt.gca()##gca='get current axis'
ax.spines['right'].set_color('none')##将右边坐标轴变为透明
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')
ax.spines['bottom'].set_position(('data',0))##以data做参照设置轴的位置
ax.spines['left'].set_position(('data',0))

for label in ax.get_xticklabels() + ax.get_yticklabels():
    label.set_fontsize(12)
    label.set_zorder(2)
    label.set_bbox(dict(facecolor='white',edgecolor='None',alpha=0.7))

plt.show()
```
