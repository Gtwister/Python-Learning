## 饼状图
1.饼状图显示一个数据系列中各项的大小与各项总和的比列。

plt.pie()

```python
import matplotlib.pyplot as plt

labels='Frogs','Hogs','Dogs','Logs'

fracs=[15,30,45,10]
colors = ['yellowgreen','gold','lightskyblue','lightcoral']
explode =(0,0.1,0,0)

plt.pie(x=fracs,labels=labels,colors=colors,explode=explode,autopct='%1.1f%%',shadow=True,startangle=90)
plt.axis('equal')

plt.show()
```
