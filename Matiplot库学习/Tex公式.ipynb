## Tex 公式

matpolotlib自带mathtext引擎，不需安装Tex系统\
$作为开始或者结束符，$ 如 '$ y=x+z $'

```python
import matplotlib.pyplot as plt
import numpy as np

fig=plt.figure()

ax=fig.add_subplot(111)

ax.set_xlim([1,7])

ax.set_ylim([1,5])

ax.text(2,4,r"$ \alpha_i \beta_j \pi \lambda \omega $",size=15)## r''表示内容不转义，\n就是转义字符，作为空行

ax.text(4,4,r"$ sin(0)=cos(\frac{\pi}{2}) $",size=15)

ax.text(2,2,r"$\lim_{x \rightarrow y} \frac{1}{x^3}$",size=15)

ax.text(4,2,r"$\sqrt[4]{x}=\sqrt{y}$",size=15)
plt.show()
```
