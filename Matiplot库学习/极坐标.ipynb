## 极坐标

```python
import numpy as np
import matplotlib.pyplot as plt

r=np.ones(9)*5

theta=np.linspace(0,2*np.pi,9)

ax=plt.subplot(111,projection='polar')## projection投影,'polar'投影成极坐标

ax.plot(theta,r,color='r',linewidth=3)

ax.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

r=np.arange(1,6,1)

theta=[0,np.pi/2,np.pi,3*np.pi/2,2*np.pi]

ax=plt.subplot(111,projection='polar')## projection投影,'polar'投影成极坐标

ax.plot(theta,r,color='r',linewidth=3)

ax.grid(True)

plt.show()
```
