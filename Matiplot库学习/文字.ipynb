## 文字

`plt.text(x,y,string,fontsize,verticalalignment,horizontalalignment)`

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-10,11,1)

y = x*x

plt.plot(x,y)

plt.text(-2,40,'function:y=x*x',family='serif',fontsize=15,color='r',style='italic',weight=1000)## family=字体类型 style=文字类型（斜体等）

plt.text(-2,20,'function:y=x*x',family='fantasy',fontsize=15,color='g',bbox=dict(facecolor='r',alpha=0.2))
plt.show()
```
