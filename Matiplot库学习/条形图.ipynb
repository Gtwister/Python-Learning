## 条形图

plt.bar()\
作用:画出条形图

```python
import numpy as np
import matplotlib.pyplot as plt

N=5

y=[20,10,20,25,15]

index = ['paris','beijing','shanghai','nanjing','nankang']

p1 = plt.bar(x=index,height=y,color='r',width=0.5)

plt.show()
```

画出水平的条形图

```python
import numpy as np
import matplotlib.pyplot as plt

N=5

y=[20,10,20,25,15]

index = ['paris','beijing','shanghai','nanjing','nankang']
## methond 1
p1 = plt.bar(x=0,bottom=index,width=y,color='r',height=0.5,orientation='horizontal')##oreintation='horizontal'转化成水平的条形图,bottom参数作为纵坐标
##############################width=就是横过来的长度,height=条形图横过来的长度
plt.show()
```

并列条形图

```python
import numpy as np
import matplotlib.pyplot as plt
index = np.arange(4)
sales_BJ=[52,55,63,53]
sales_SH=[44,66,55,41]

bar_width=0.3

plt.bar(index,sales_BJ,bar_width,color='b')
plt.bar(index+bar_width,sales_SH,bar_width,color='r')

plt.show()
```

层叠条形图

```python
import numpy as np
import matplotlib.pyplot as plt
index = np.arange(4)
sales_BJ=[52,55,63,53]
sales_SH=[44,66,55,41]

bar_width=0.3

plt.bar(index,sales_BJ,bar_width,color='b')
plt.bar(index,sales_SH,bar_width,color='r',bottom=sales_BJ)

plt.show()
```
