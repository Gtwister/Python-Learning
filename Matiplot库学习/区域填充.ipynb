## 区域填充

fill(),fill_between()

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,5*np.pi,1000)

y1=np.sin(x)

y2=np.sin(2*x)

plt.fill(x,y1,'b',alpha=0.3)

plt.fill(x,y2,'r',alpha=0.3)
plt.grid(color='b',linewidth='0.1',alpha=0.5)
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,5*np.pi,100)

y1=np.sin(x)

y2=np.sin(2*x)

plt.plot(x,y1,color='r')
plt.plot(x,y2,color='b')
plt.fill_between(x,y1,y2,where=y1>y2,facecolor='blue',interpolate=True)##当自变量比较离散时，用interpolate=True自动填充
plt.fill_between(x,y1,y2,where=y2>y1,facecolor='green',interpolate=True)
plt.show()
```
