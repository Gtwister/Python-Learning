## 函数积分图

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

def fun(x):
    return -(x-2)*(x-8)+10
x=np.linspace(0,10)

y=fun(x)

fig,ax=plt.subplots()

plt.plot(x,y,'r',linewidth=2)
a=2
b=9
ax.set_xticks([a,b],[r'$a$',r'$b$'])
ax.set_yticks([])

ix=np.linspace(a,b)
iy=fun(ix)
plt.ylim(0)
verts=[(a,0),*zip(ix,iy),(b,0)]

poly=Polygon(verts,facecolor='0.9',edgecolor='0.5')
ax.add_patch(poly)
plt.figtext(0.9,0.07,r'$x$')
plt.figtext(0.1,0.87,r'$y$')
x_math=(a+b)*0.5-2.5
y_math=7
plt.text(x_math,y_math,r"$\int_a^b (-(x-2)*(x-8)+10)dx$",fontsize=12)
plt.show()
```
