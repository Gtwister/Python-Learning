## 直方图
1.表示数据分布的情况\
2.与条形图的区别：条形图反映的是不同类别的数据情况，直方图表示的是连续数据的情况

```python
import numpy as np
import matplotlib.pyplot as plt

mu = 100 # mean of distribution
sigma = 20 # standard deviation of distribution
x = mu + sigma*np.random.randn(200000)

plt.hist(x,bins=100,color='green',edgecolor='black',density=True)## x=输入的变量, bins=生成的柱状图数量, color=图形颜色, normed=布尔值，是否标准化

plt.show()
```
