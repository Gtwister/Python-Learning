## 坐标轴范围

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-10,11,1)

plt.plot(x,x*x)

plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-10,11,1)

plt.plot(x,x*x)
plt.axis([-5,5,0,60])##axis([xmin,xmax,ymin,ymax])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-10,11,1)

plt.plot(x,x*x)
plt.xlim([-5,5])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-10,11,1)

plt.plot(x,x*x)
plt.ylim([0,60])
plt.show()
```
