### Tensor常见的形式有哪些

* 0:scalar
* 1:vector
* 2:matrix
* 3:n-dimensional tensor

```python
import torch
from torch import tensor
```

### Scalar
通常是一个数值

```python
x = tensor(42.)
x
```

```python
x.dim()
```

```python
x.item()
```

### Vector

例如: [-5.,2.,0.],在深度学习中通常指特征，例如词向量特征，某一维度特征等

```python
v = tensor([1.5, -0.5, 3.0])
v
```

```python
v.dim()
```

```python
v.size()
```

### Matrix

* 一般计算的都是矩阵，通常是多维的

```python
M = tensor([[1.,2.],[3.,4.]])
M
```

```python
M.matmul(M)
```

```python
tensor([1.,0.]).matmul(M)
```

```python
M*M
```
