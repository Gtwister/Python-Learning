### 举个例子:

![alt text](pic/backward.drawio.png)

y = w*x
z = y + b
f(x) = w*x + b
模型是一元一次的
初始时我就已经有了一个稳定的输入b
我们每此对这个模型的输入是x
1.5x, 0.8x

```python
#计算流程
x = torch.rand(1)
w = torch.rand(1,requires_grad=True)
b = torch.rand(1,requires_grad=True) 

y=x*w
z=y+b

```

```python
x.requires_grad,b.requires_grad,w.requires_grad,y.requires_grad,z.requires_grad#注意y也是需要的,因为y是通过w*x得到的结果
```

```python
x.is_leaf,w.is_leaf,b.is_leaf,y.is_leaf,z.is_leaf
```

### 反向传播计算

```python
z.backward(retain_graph=True)#如果不清空反向求导的梯度会累加在一起， grad = grad + new_grad 
z.backward()
```
