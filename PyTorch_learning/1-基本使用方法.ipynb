```python
import torch
```

创建一个随机值，矩阵

```python
x = torch.rand(5,3)
x
```

初始化一个全零矩阵

```python
x = torch.zeros(5,3, dtype=torch.long)
x
```

直接传入数据

```python
x = torch.tensor([5.5, 3])
x
```

```python
x = x.new_ones(5,3,dtype=torch.double)

x = torch.randn_like(x, dtype=torch.float)
x
```

展示矩阵大小

```python
x.size()
```

### view操作可以改变矩阵维度

```python
import torch
```

```python
x = torch.tensor([1,2,3,4])
x
```

```python
x = torch.randn(4,4)

y = x.view(16)

z = x.view(-1,8) ## -1是自动计算的意思

print(x.size(),y.size(),z.size())
```

### 与Numpy协同操作

```python
a = torch.ones(5)
b = a.numpy()
b
```

```python
import numpy as np
a = np.ones(5)
b = torch.from_numpy(a)
b
```
