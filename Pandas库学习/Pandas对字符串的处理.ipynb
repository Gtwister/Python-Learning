## 对字符串的处理

**Pandas的字符串处理:**
* 1.使用方法:先获取Series的str属性，然后在属性上调用函数;
* 2.只能在字符串列上使用,不能在数字列上使用;
* 3.Dataframe上没有str属性和处理方法
* 4.Series.str并不是Py的原生字符串，而是自己的一套方法，不过大部分和原生很相似;

**Series.str相关网址**\
(https://pandas.pydata.org/docs/reference/series.html?highlight=string%20handling)

```python
import pandas as pd
```

```python
fpath = './beijing_tianqi_2018.csv'
df = pd.read_csv(fpath)
```

**1、获取Series的str属性,使用各种字符串处理函数**

```python
df['bWendu'].str
```

```python
# 字符串替换函数
df['bWendu'].str.replace("℃","")
```

```python
# 判断是不是数字
df['bWendu'].str.isnumeric()
```

```python
df['bWendu'].str.len()
```

**2、使用str的startwith、contains等得到bool的Series可以做条件查询**

```python
condition = df['ymd'].str.startswith('2018-03')
df[condition].head()
```

**3、需要多次str处理的链式操作**


```python
df['ymd'].str.replace('-','')##返回的是Series属性
```

```python
df['ymd'].str.replace('-','').slice(0,6)##不能直接使用slice切片操作对Series
```

```python
df['ymd'].str.replace('-','').str.slice(0,6)
```

**4.使用正则表达式的处理**

```python
#添加新列
def get_nianyueri (x):
    year,month,day = x['ymd'].split('-')##使用x['ymd]得到的式字符串，所以不是Series，可以使用split分割
    return f'{year}年{month}月{day}日'
df['中文日期'] = df.apply(get_nianyueri,axis=1)
```

```python
df.head()
```

怎样去除中文'年','月','日'

```python
##方法1: 链式replace
df['中文日期'].str.replace('年','').str.replace('月','').str.replace('日','')
```

```python
#方法2: 正则表达式替换
df['中文日期'].str.replace('[年月日]','')
```
