## 数据结构

### 1. Series
Series 是一种类似一维数组的对象、它由一组数据（不同数据类型）以及一组与之相关的数据标签（即索引）组成。

#### 1.1仅有的数据列表可产生最简单的Series

```python
import pandas as pd
import numpy as np
```

```python
s1 = pd.Series([1,'a',5.2,7])

# 左侧为索引，右侧是数据
s1
```

```python
# 获取索引
s1.index
```

```python
# 获取数据
s1.values
```

#### 1.2创建一个具有标签索引的Series

```python
s2 = pd.Series([1,'a',5.2,3],index=['a','b','c','d'])

s2
```

#### 1.3用python字典创建Series

```python
sdata={'Ohio':35000,'Texas':72000,'Oregon':16000,'Utah':5000}

s3 = pd.Series(sdata)

s3
```

#### 1.4根据标签索引查询数据

```python
s2
```

```python
s2['a']
```

```python
s2[['a','b']]##查询多值
```

# DataFrame

DataFrame 是一个表格型的数据结构
* 每列可以是不同的值类型
* 既有行索引index,也有列索引columns
* 可以被看做由Series组成的字典 

```python
data ={
    'state':['Ohio','Ohio','Ohio','Nevada','Nevada'],
    'year':[2000,2001,2002,2001,2002],
    'pop':[1.5,1.7,3.6,2.4,2.9]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df.dtypes
```

```python
df.columns
```

```python
df.index
```

### 3.1 查询一列返回的是一个Series对象

```python
df[['year']]
```

**3.2查询多列返回的是dataframe对象**

```python
df[['year','pop']]
```

**3.3查询行**

```python
df.loc[1]
```
