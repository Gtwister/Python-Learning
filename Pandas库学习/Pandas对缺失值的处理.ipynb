## 对缺失值的处理

Pandas使用这些函数处理缺失值:

* isnull和notnull: 检测是否是空值，可用于df和series

* dropna: 丢弃、删除缺失值
  - axis:删除行还是列,{0 or 'index',1 or 'columns'}
  - how:如果等于any则任何空值都会被删除，如果等于all则所有值为空才删除
  - inplace:如果为True则修改当前df,否则返回新的df
  
  
* fillna：填充空值

  - value:用于填充的值，可以是单个值，或者是字典。
  - method: 等于ffill使用前一个不为空的值填充forword fill; 等于bfill使用后一个不为空的值填充的backward fill
  - axis: 按行还是列填充
  - inplace: 如果为True则修改当前df,否则返回新的df 

```python
import pandas as pd
```

```python
studf = pd.read_excel('./students_scores.xlsx',skiprows=2)
studf
```

**step4:删除掉全是空值的行**

```python
studf.dropna(axis='index',how='all',inplace=True)
```

```python
studf
```

**step5:将分数列为空的填充为0分**

```python
studf.fillna({'分数':0})
```

```python
# 等同于
studf.loc[:,'分数'] = studf['分数'].fillna(0)
```

```python
studf
```

**step6:将姓名的缺失值填充**\
使用前面的有效值填充,用ffill: forward fill

```python
studf.loc[:,"姓名"] = studf['姓名'].fillna(method='ffill')
studf
```

**step7:将清洗好的excel保存**

```python
studf.to_excel('./students_scores.xlsx',index=False)
```
